```python
import requests as r
import json
url ='https://services.odata.org/V4/Northwind/Northwind.svc/Orders?$filter=ShipCountry%20eq%20%27Germany%27%20and%20Freight%20gt%2020&$format=json'
response = r.get(url)
```

```python
import yaml
from langchain_community.agent_toolkits import JsonToolkit, create_json_agent
from langchain_community.tools.json.tool import JsonSpec
from langchain_openai import OpenAI
```

```python
result = response.json()
data=result['value']
type(result)
```

```python

json_spec = JsonSpec(dict_=result, max_value_length=10000)
json_toolkit = JsonToolkit(spec=json_spec)

json_agent_executor = create_json_agent(
    llm=OpenAI(temperature=0), toolkit=json_toolkit, verbose=True
)
```

```python
json_agent_executor.invoke(
    "How many orders sent to germany have freight cost greater than 100 $?"
)
```
