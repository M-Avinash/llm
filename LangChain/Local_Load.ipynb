```python
from langchain_community.llms import LlamaCpp
```

```python
n_gpu_layers = -1  # Metal set to 1 is enough.
n_batch = 1024  # Should be between 1 and n_ctx, consider the amount of RAM of your Apple Silicon Chip.

# Make sure the model path is correct for your system!
llm = LlamaCpp(
    model_path="/home/avinash/Documents/Work/LLM Applications/Models/gemma-2-27b-it-Q4_K_M.gguf",
    n_gpu_layers=n_gpu_layers,
    n_batch=n_batch,
    n_ctx=2048,
    f16_kv=True,  # MUST set to True, otherwise you will run into problem after a couple of calls
    verbose=True,
)
```

```python
result=llm.invoke("Give me the lyrics for - Lord Finesse - Hip 2 Da Game song")
```
