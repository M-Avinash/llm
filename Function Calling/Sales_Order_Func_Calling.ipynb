```python
import requests as r
import json

# Specify the LLM Endpoint
API_URL = "http://nexusraven.nexusflow.ai"
headers = {
    "Content-Type": "application/json"
}

def get_northwind_data_details(Category: str, ID: str):
    """
    Get Categories. You can collect Order with its Order ID or Product with its Product ID.
    """
    url = f"https://services.odata.org/V2/Northwind/Northwind.svc/{Category}({ID})?$format=json"
    response = r.get(url)
    print(response.json())

def query_raven(prompt):
    """
    This function sends a request to the TGI endpoint to get Raven's function call.
    This will not generate Raven's justification and reasoning for the call, to save on latency.
    """
    output = raven_post({
        "inputs": prompt,
        "parameters": {
            "temperature": 0.001,
            "stop": ["<bot_end>"],
            "do_sample": False,
            "max_new_tokens": 2048,
            "return_full_text": False
        }
    })
    call = output[0]["generated_text"].replace("Call:", "").strip()
    return call

def raven_post(payload):
    """
    Sends a payload to a TGI endpoint.
    """
    import requests
    response = requests.post(API_URL, headers=headers, json=payload)
    return response.json()

USER_QUERY = "from Orders get details of  order Id 10259"

raven_functions = f'''
import requests as r

def get_northwind_data_details(Category : str,  ID : str):
    """
    Get Categories. You can collect Order with its Order ID or Product with its Product ID.
    """
    url  =   f"https://services.odata.org/V2/Northwind/Northwind.svc/{{Category}}({{ID}})?$format=json"
    response = r.get(url)
    print(response.json())

# User Query: {USER_QUERY}<human_end>
'''

# Execute Raven query and store the result
result = {}
exec(raven_functions, globals(), result)

# Now execute the generated code from Raven
call = query_raven(raven_functions)
exec(call, globals(), result)

# Access any variables or outputs from the executed code
print(result)
```
